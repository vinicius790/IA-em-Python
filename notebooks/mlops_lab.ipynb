# Iris AI Platform · MLOps Lab
Notebook demonstrativo para profiling, qualidade, treinamento, benchmarking e monitoramento.

```python
from app.data import load_iris_dataframe
from app.data_catalog.profiling import profile_dataframe
from app.ml import build_models, train_model
from app.quality import validate_training_frame

df = load_iris_dataframe()
df.head()
```

```python
profile = profile_dataframe(df)
quality = validate_training_frame(df)
print(profile)
print("quality:", quality.passed, quality.score)
```

```python
# Benchmark rápido: todos os modelos disponíveis no ambiente.
from pathlib import Path

results = []
for name in build_models():
    r = train_model(df, name, Path("artifacts/notebook"), cv_folds=3)
    results.append((name, r.metrics["accuracy"], r.metrics["f1_macro"]))
sorted(results, key=lambda x: x[2], reverse=True)[:10]
```

```python
# Tuning do XGBoost: Optuna com o extra `tuning`; sem ele, RandomizedSearchCV.
optimized = train_model(
    df,
    "xgboost",
    Path("artifacts/notebook-optuna"),
    cv_folds=3,
    tune_hyperparameters=True,
    optimization_trials=6,
)
optimized.metrics["optimization"]
```
